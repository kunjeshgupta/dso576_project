```python
def extract_listen(listen, username):
    metadata = listen.get("track_metadata", {})
    additional = metadata.get("additional_info", {})
    mapping = metadata.get("mbid_mapping", {})

    return {
        "user_name": username,
        "listened_at": listen.get("listened_at"),
        "recording_msid": listen.get("recording_msid"),
        "artist_name": metadata.get("artist_name"),
        "track_name": metadata.get("track_name"),
        "release_name": metadata.get("release_name"),

        # Prefer server-resolved MusicBrainz IDs and use submitted IDs as fallback
        "artist_mbids": (
            mapping.get("artist_mbids")
            or additional.get("artist_mbids")
        ),
        "recording_mbid": (
            mapping.get("recording_mbid")
            or additional.get("recording_mbid")
        ),
        "release_mbid": (
            mapping.get("release_mbid")
            or additional.get("release_mbid")
        ),
        "release_group_mbid": (
            mapping.get("release_group_mbid")
            or additional.get("release_group_mbid")
        ),

        "duration_ms": additional.get("duration_ms"),
        "music_service": additional.get("music_service"),
        "submission_client": additional.get("submission_client")
    }
```

```python
rows = [extract_listen(listen, username) for listen in listens]

df = pd.DataFrame(rows)

df.head()
```

```python
df["listened_datetime"] = pd.to_datetime(
    df["listened_at"],
    unit="s",
    utc=True
)

df["listen_date"] = df["listened_datetime"].dt.date

df.head()
```

```python
import time


def fetch_user_listens(username, start_date, end_date):
    start_ts = int(pd.Timestamp(start_date, tz="UTC").timestamp())

    # Add one day so the complete end date is included
    end_exclusive = pd.Timestamp(end_date, tz="UTC") + pd.Timedelta(days=1)
    cursor = int(end_exclusive.timestamp())

    all_listens = []

    while True:
        url = f"https://api.listenbrainz.org/1/user/{username}/listens"

        response = requests.get(
            url,
            params={
                "count": 1000,
                "max_ts": cursor
            },
            headers={"User-Agent": "dso576-student-project/0.1"},
            timeout=30
        )

        response.raise_for_status()
        batch = response.json()["payload"]["listens"]

        if not batch:
            break

        timestamps = [
            listen["listened_at"]
            for listen in batch
            if listen.get("listened_at") is not None
        ]

        if not timestamps:
            break

        for listen in batch:
            timestamp = listen.get("listened_at")

            if timestamp is not None and start_ts <= timestamp < int(end_exclusive.timestamp()):
                all_listens.append(extract_listen(listen, username))

        oldest_timestamp = min(timestamps)

        if oldest_timestamp < start_ts:
            break

        cursor = oldest_timestamp
        time.sleep(0.2)

    return pd.DataFrame(all_listens)
```

```python
df_rob = fetch_user_listens(
    username="rob",
    start_date="2026-06-01",
    end_date="2026-08-31"
)

print(df_rob.shape)
df_rob.head()
```

```python
df_rob["listened_datetime"] = pd.to_datetime(
    df_rob["listened_at"],
    unit="s",
    utc=True
)

df_rob["listen_date"] = df_rob["listened_datetime"].dt.date
df_rob["listen_month"] = df_rob["listened_datetime"].dt.to_period("M").astype(str)
```

```python
from pathlib import Path

Path("../data").mkdir(exist_ok=True)

df_rob.to_json(
    "../data/listenbrainz_sample.jsonl",
    orient="records",
    lines=True
)
```
